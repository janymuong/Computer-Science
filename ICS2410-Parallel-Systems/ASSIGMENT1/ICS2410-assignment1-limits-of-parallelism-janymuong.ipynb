Amdahl’s Law – Speedup Computation:
=========

Amdahl’s Law gives the theoretical speedup of a fixed‑size task when using \(N\) processors:

$$
\text{Speedup}(N) = \frac{1}{(1-P) + \frac{P}{N}}
$$

where \(P\) = fraction of code that can be parallelized, and \(1-P\) = serial fraction.



### PSET 1  
\(P = 0.60,\; N = 10\)  

$$
\begin{aligned}
\text{Speedup} &= \frac{1}{(1-0.60) + \frac{0.60}{10}} \\
&= \frac{1}{0.40 + 0.06} = \frac{1}{0.46} \approx 2.17
\end{aligned}
$$

### PSET 2  
\(P = 0.90,\; N = 10\)  

$$
\begin{aligned}
\text{Speedup} &= \frac{1}{(1-0.90) + \frac{0.90}{10}} \\
&= \frac{1}{0.10 + 0.09} = \frac{1}{0.19} \approx 5.26
\end{aligned}
$$

### PSET 3  
\(P = 0.99,\; N = 10\)  

$$
\begin{aligned}
\text{Speedup} &= \frac{1}{(1-0.99) + \frac{0.99}{10}} \\
&= \frac{1}{0.01 + 0.099} = \frac{1}{0.109} \approx 9.17
\end{aligned}
$$

#### Amdahl's Table


Graph for Amdahl’s Law:
=========
> I am using Python/Matplotlib/Numpy for plotting

```python
import matplotlib.pyplot as plt
import numpy as np

N = [2, 4, 8, 16, 32, 64, 128, 256]
P_values = [0.50, 0.75, 0.90, 0.95]
colors = ['blue', 'green', 'orange', 'red']

plt.figure(figsize=(8, 5))
for P, c in zip(P_values, colors):
    speedup = [1 / ((1 - P) + P / n) for n in N]
    plt.plot(N, speedup, marker='o', linestyle='-', color=c, label=f'P = {P*100:.0f}%')

plt.xscale('log')
plt.xlabel('Number of Processors (N)')
plt.ylabel('Speedup')
plt.title('Amdahl\'s Law: Speedup vs. Number of Processors')
plt.grid(True, which='both', linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig('amdahl_graph.png', dpi=300)
plt.show()
```

**Explanation of the diagram:**  
Each curve corresponds to a different parallel fraction. The curve for \(P = 0.50\) barely exceeds 2× speedup, because half the code is serial. For \(P = 0.95\), speedup reaches about 18.6× with 256 processors, but it is far from ideal linear speedup. The flattening of the curves shows the diminishing returns predicted by Amdahl’s Law – the serial part eventually dominates performance.

Gustafson–Barsis’s Law
=========
> Calculations and graph of it:

Gustafson’s Law applies to variable‑sized problems where the problem scales with the number of processors. Speedup is given by:

$$
S(N) = N + (1 - N) \times s
$$

where \(s\) is the serial fraction (\(s = 1 - P\)). We use the same \(P\) values and \(N\) range.

**For \(P = 0.50\)** (\(s = 0.50\)):

$$
S(N) = N + (1 - N) \times 0.50 = N + 0.5 - 0.5N = 0.5 + 0.5N
$$

- \(N=2\): \(0.5 + 0.5 \times 2 = 0.5 + 1.0 = 1.5\)
- \(N=4\): \(0.5 + 0.5 \times 4 = 0.5 + 2.0 = 2.5\)
- \(N=8\): \(0.5 + 0.5 \times 8 = 0.5 + 4.0 = 4.5\)
- \(N=16\): \(0.5 + 0.5 \times 16 = 0.5 + 8.0 = 8.5\)
- \(N=32\): \(0.5 + 0.5 \times 32 = 0.5 + 16.0 = 16.5\)
- \(N=64\): \(0.5 + 0.5 \times 64 = 0.5 + 32.0 = 32.5\)
- \(N=128\): \(0.5 + 0.5 \times 128 = 0.5 + 64.0 = 64.5\)
- \(N=256\): \(0.5 + 0.5 \times 256 = 0.5 + 128.0 = 128.5\)

**For \(P = 0.75\)** (\(s = 0.25\)):

$$
S(N) = N + (1 - N) \times 0.25 = N + 0.25 - 0.25N = 0.25 + 0.75N
$$

- \(N=2\): \(0.25 + 0.75 \times 2 = 0.25 + 1.5 = 1.75\)
- \(N=4\): \(0.25 + 0.75 \times 4 = 0.25 + 3.0 = 3.25\)

- \(N=16\): \(0.25 + 0.75 \times 16 = 0.25 + 12.0 = 12.25\)
- \(N=32\): \(0.25 + 0.75 \times 32 = 0.25 + 24.0 = 24.25\)

- \(N=128\): \(0.25 + 0.75 \times 128 = 0.25 + 96.0 = 96.25\)
- \(N=256\): \(0.25 + 0.75 \times 256 = 0.25 + 192.0 = 192.25\)

**For \(P = 0.90\)** (\(s = 0.10\)):

$$
S(N) = N + (1 - N) \times 0.10 = N + 0.1 - 0.1N = 0.1 + 0.9N
$$

- \(N=2\): \(0.1 + 0.9 \times 2 = 0.1 + 1.8 = 1.9\)
- \(N=4\): \(0.1 + 0.9 \times 4 = 0.1 + 3.6 = 3.7\)

**For \(P = 0.95\)** (\(s = 0.05\)):

- \(N=2\): \(0.05 + 0.95 \times 2 = 0.05 + 1.9 = 1.95\)

### Gustafson Table


```python
import matplotlib.pyplot as plt
import numpy as np

N = [2, 4, 8, 16, 32, 64, 128, 256]
P_values = [0.50, 0.75, 0.90, 0.95]
colors = ['blue', 'green', 'orange', 'red']

plt.figure(figsize=(8, 5))
for P, c in zip(P_values, colors):
    s = 1 - P
    speedup = [n + (1 - n) * s for n in N]
    plt.plot(N, speedup, marker='o', linestyle='-', color=c, label=f'P = {P*100:.0f}%')

plt.plot(N, N, linestyle='--', color='gray', label='Ideal linear speedup')
plt.xlabel('Number of Processors (N)')
plt.ylabel('Speedup')
plt.title('Gustafson–Barsis’s Law: Speedup vs. Number of Processors')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig('gustafson_graph.png', dpi=300)
plt.show()
```

**Explanation of Gustafson’s diagram:**  
The curves are close to the ideal linear speedup line (dashed). For a scalable problem, increasing the number of processors allows us to solve a proportionally larger problem, so the serial fraction becomes less restrictive. The slope of each curve is determined by the serial fraction \(s\): a smaller serial fraction yields a steeper slope, approaching the ideal line. This law better models many real‑world applications where users scale the problem size with available resources.
