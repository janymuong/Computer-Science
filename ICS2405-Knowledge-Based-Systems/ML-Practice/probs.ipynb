### [1.1] Binomial Distribution
If I flip a coin many, many times and count the number of heads, here’s what happens as we increase the number of flips:

```python
import scipy.stats as ss

binomal_flips = [5, 10, 20, 40, 80]

b = ss.distributions.binom
for flips in binomal_flips:
    # binom wit a .5 is a result of too many coin flips
    success = np.arange(flips)
    distro = b.pmf(success, flips, .5)
    plt.hist(success, flips, weights=distro)

plt.xlim(0, 55)
```

My mathematical
colleagues may cringe, but the primary feature we need from the normal distribution is its
shape

f I ignore that the whole numbers are counts and replace the graph with a smthoother
curve that takes values everywhere, instead of the stair steps that climb or descend at whole
numbers, I get something like this:

```python
DISTRO_FLIPS = [5, 10, 20, 40, 80]

b = ss.distributions.binom
n = ss.distributions.norm

for flips in DISTRO_FLIPS:
    # binomial coins
    success = np.arange(flips)
    our_distro = b.pmf(success, flips, 0.5)
    plt.hist(success, flips, weights=our_distro)

    # normal approximatiom to the binomial;
    # set the mean and std deviation

    mu = flips * .5
    std_dev = np.sqrt(flips * .5 * (1 - .5))

    # we have to set up both the x and y points for the normal
    # we get the ys from the distribution (a function)
    # we have to feed it xs, we set those up here
    norm_x = np.linspace(mu-3*std_dev, mu+3*std_dev, 100)
    norm_y = n.pdf(norm_x, mu, std_dev)
    plt.plot(norm_x, norm_y, 'k');
plt.xlim(0, 55);
    
```
